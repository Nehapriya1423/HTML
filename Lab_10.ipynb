```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    api_key="gsk_API_KEY",
    api_secret="gsk_API_SECRET
    model="llama-3.1-8b-instant",
    temperature=0.3
)

print("Groq model connected successfully!")
```

```python
from langchain_community.tools import DuckDuckGoSearchResults

search = DuckDuckGoSearchResults()

def search_agent(topic):
    return search.run(f"latest news about {topic}")

print("Search Agent created successfully!")
```

```python
def summariser_agent(raw_results):
    return llm.invoke(
        "Summarise these news results into 3 short bullet points:\n"
        + raw_results
    ).content

print("Summariser Agent created successfully!")
```

```python
def report_agent(topic, summary):
    return llm.invoke(
        f'Create a short Markdown report titled "News: {topic}". '
        f'Use a heading and these points:\n{summary}'
    ).content

print("Report Agent created successfully!")
```

```python
topic = "artificial intelligence"

raw = search_agent(topic)          # Agent 1

summary = summariser_agent(raw)    # Agent 2

report = report_agent(topic, summary)  # Agent 3


with open("news_report.md", "w") as f:
    f.write(report)


print(report)

print("\nSaved as news_report.md (see the Files panel on the left).")
```
