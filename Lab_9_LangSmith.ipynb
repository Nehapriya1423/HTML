```python
import os

os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_API_KEY"] = "lsv2_API_KEY"
os.environ["LANGCHAIN_PROJECT"] = "Lab9-Agentic-AI"

print("LangSmith tracing enabled successfully!")
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    api_key="gsk_API_KEY",
    model="llama-3.1-8b-instant",
    temperature=0
)

question = (
    "A shop sells pens at 12 rupees each. "
    "If I buy 7 pens and pay with 100 rupees, "
    "how much change will I get? Think step by step."
)

response = llm.invoke(question)

print(response.content)
```
