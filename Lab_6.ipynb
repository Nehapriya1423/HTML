```python
import autogen
import openai

print("AutoGen version:", autogen.__version__)
print("OpenAI version:", openai.__version__)
```

```python
config_list = [
    {
        "model": "llama-3.1-8b-instant",
        "api_key": "gsk_API_KEY",
        "base_url": "https://api.groq.com/openai/v1"
    }
]

llm_config = {
    "config_list": config_list,
    "temperature": 0.4
}

print("Groq model configured successfully!")
```

```python
!pip uninstall -y httpx
!pip install -q httpx==0.27.0
```

```python
writer = autogen.AssistantAgent(
    name="Writer",
    system_message=(
        "You write short, clear blog posts. "
        "When given feedback, rewrite to improve it."
    ),
    llm_config=llm_config
)

print("Writer agent created successfully!")
```

```python
critic = autogen.AssistantAgent(
    name="Critic",
    system_message=(
        "You review blog posts and give 2-3 specific improvements. "
        "If it is good enough, reply APPROVED."
    ),
    llm_config=llm_config
)

print("Critic agent created successfully!")
```

```python
critic.initiate_chat(
    writer,
    message="Write a 150-word blog post about the benefits of reading.",
    max_turns=4
)
```
