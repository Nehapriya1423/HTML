```python
from langchain_groq import ChatGroq
from langchain.agents import initialize_agent, AgentType, Tool
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

print("Imports successful!")
```

```python
llm = ChatGroq(
    api_key="GROQ_API_KEY",
    model="llama-3.1-8b-instant",
    temperature=0
)

print("LLM created successfully!")
```

```python
def calculator(expression: str):
    try:
        allowed = set("0123456789+-*/(). ")
        if not set(expression).issubset(allowed):
            return "Error: Only numbers and + - * / ( ) are allowed."
        return str(eval(expression))
    except Exception as e:
        return f"Error: {e}"

print("Calculator created!")
```

```python
wiki = WikipediaQueryRun(
    api_wrapper=WikipediaAPIWrapper()
)

print("Wikipedia tool created!")
```

```python
tools = [
    Tool(
        name="Wikipedia",
        func=wiki.run,
        description="Look up facts about people, places, and events."
    ),
    Tool(
        name="Calculator",
        func=calculator,
        description="Perform arithmetic calculations."
    )
]

print("Tools registered!")
```

```python
agent = initialize_agent(
    tools,
    llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

print("Agent created successfully!")
```

```python
question = (
    "Who developed the theory of relativity, "
    "and what is his birth year multiplied by 2?"
)

answer = agent.invoke({"input": question})

print("\nFINAL ANSWER:", answer["output"])
```
